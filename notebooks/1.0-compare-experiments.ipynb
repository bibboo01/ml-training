# 1.0 Compare experiments from MLflow
Read all runs of an experiment and compare test metrics, incl. TP/FP/FN/TN.

```python
import mlflow
from ml_training.utils import PROJECT_ROOT

mlflow.set_tracking_uri(f"sqlite:///{(PROJECT_ROOT / 'mlflow.db').as_posix()}")
runs = mlflow.search_runs(experiment_names=["breast_cancer"], filter_string="tags.stage = 'train'")
cols = ["tags.model", "metrics.test_accuracy", "metrics.test_f1", "metrics.test_roc_auc",
        "metrics.test_TP", "metrics.test_FP", "metrics.test_FN", "metrics.test_TN"]
runs[cols].sort_values("metrics.test_f1", ascending=False)
```
